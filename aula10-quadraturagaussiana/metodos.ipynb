```python
def quadratura_gaussiana(metodo, N, funcaoDada):
    f = funcaoDada
    
    # N -> número de termos do somatório do método da quadratura (N = n + 1), como 2n + 1 = 3 (grau) -> n = 1 -> N = 2
    
    soma = 0

    A, x = tabela_quadratura_gaussiana(metodo, N)

    for i in range (0, N):
        soma += A[i] * f(x[i])

    I_quadratura = soma

    print(f"Resultado da Integral: {I_quadratura}")

    return I_quadratura
```
