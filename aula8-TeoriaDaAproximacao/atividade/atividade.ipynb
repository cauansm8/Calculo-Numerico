## 1

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xi = np.array([-1., 0, 1, 2])
yi = np.array([0., -1, 0, 7])
```

```python
def minimosQuadrados(grau = 2):

    m = grau

    # número de pontos
    n = len(xi)

    # número de coeficientes do polinômio
    mn = m + 1

    u = np.zeros((n, mn))

    for i in range (0, mn):
        u[:, i] = xi ** i

    # número de pontos para curva
    N = 100

    x = np.linspace(xi[0], xi[n-1], N)

    G = np.zeros((mn, mn))
    b = np.zeros(mn)

    for i in range (0, mn):
        b[i] = np.sum(u[:, i] * yi)

        for j in range (0, mn):
            G[i,j] = np.sum(u[:, i] * u[:, j])


    # resolver o sistema linear
    a = np.linalg.inv(G) @ b


    # erro de truncamento Q
    nos = u @ a
    Q = np.sum((yi[:] - nos) ** 2)
    print(f"Erro de Truncamento: {Q}")



    P = np.zeros((N, mn))

    for i in range(0, mn):
        P[:, i] = x[:] ** (i)

    y_aprox = P @ a



    return x, y_aprox, a
```

```python
# A
x, y, a, = minimosQuadrados(1)

print(f"Reta: {a[0]:.2f} + {a[1]:.2f}x")
```

```python
# B
x, y, a = minimosQuadrados(3)

print(f"Polinômio de grau 3: {a[0]:.2f} + {a[1]:.2f}x + {a[2]:.2f}x^2 + {a[3]:.2f}x^3")
```

```python
# C
x1, y1, a_1 = minimosQuadrados(1)
x2, y2, a_2 = minimosQuadrados(2)
x3, y3, a_3 = minimosQuadrados(3)


plt.plot(xi, yi, 'ro', 4)
plt.plot(x1, y1, 'b-', label='reta')
plt.plot(x2, y2, 'g-', label='^2')
plt.plot(x3, y3, 'r-', label='^3')
plt.legend()
plt.grid()
plt.show()
```

```python
# D
f = 0.2

f_reta = a_1[0] + f * a_1[1]
f_polinomio2 = a_2[0] + f * a_2[1] + (f ** 2) * a_2[2]
f_polinomio3 = a_3[0] + f * a_3[1] + (f ** 2) * a_3[2] + (f ** 3) * a_3[3]

print(f"f(0.2) na reta: {f_reta}")
print(f"f(0.2) no polinômio de grau 2: {f_polinomio2}")
print(f"f(0.2) no polinômio de grau 3: {f_polinomio3}")
```
