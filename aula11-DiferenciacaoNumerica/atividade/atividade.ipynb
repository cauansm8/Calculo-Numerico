```python
import numpy as np
from tabulate import tabulate
```

```python
def diferenciacao_numerica(x0, func, dfunc, Dfunc, delta):
    length = len(delta)
    dfunc_progressiva = np.zeros(length)
    dfunc_regressiva = np.zeros(length)
    dfunc_centrada = np.zeros(length)
    dfunc_2derivada = np.zeros(length)
    
    erro_dfprogressiva = np.zeros(length)
    erro_dfregressiva = np.zeros(length)
    erro_dfcentrada = np.zeros(length)
    erro_df2derivada = np.zeros(length)
    
    for i in range(length):
        dfunc_progressiva[i] = (func(x0 + delta[i]) - func(x0)) / delta[i]
        dfunc_regressiva[i] = (func(x0) - func(x0-delta[i])) / delta[i]
        dfunc_centrada[i] = (func(x0 + delta[i]) - func(x0-delta[i])) / (2*delta[i])
        dfunc_2derivada[i] = (func(x0 + delta[i]) - (2*func(x0)) + func(x0 - delta[i])) / (delta[i]**2)
        
        erro_dfprogressiva[i] = abs(dfunc(x0) - dfunc_progressiva[i])/dfunc(x0)
        erro_dfregressiva[i] = abs(dfunc(x0) - dfunc_regressiva[i])/dfunc(x0)
        erro_dfcentrada[i] = abs(dfunc(x0) - dfunc_centrada[i])/dfunc(x0)
        erro_df2derivada[i] = abs(Dfunc(x0) - dfunc_2derivada[i])/Dfunc(x0)
    
    retorno = [dfunc_progressiva, erro_dfprogressiva, dfunc_regressiva, erro_dfregressiva, dfunc_centrada, erro_dfcentrada, dfunc_2derivada, erro_df2derivada]
    
    return retorno
```

```python
def diferenciacao_mista(x0, y0, func, dfuncx, dfuncy, Dfuncx, Dfuncy, Dfuncxy, deltax, deltay):
    length = len(deltax)
    dfunc_progressiva_x = np.zeros(length)
    dfunc_regressiva_x = np.zeros(length)
    dfunc_centrada_x = np.zeros(length)
    dfunc_2cx = np.zeros(length)
    
    dfunc_progressiva_y = np.zeros(length)
    dfunc_regressiva_y = np.zeros(length)
    dfunc_centrada_y = np.zeros(length)
    dfunc_2centrada_y = np.zeros(length)
    
    dfunc_2centrada_xy = np.zeros(length)
    
    erro_df_progressivax = np.zeros(length)
    erro_df_regressiva_x = np.zeros(length)
    erro_df_centrada_x = np.zeros(length)
    erro_df_2x = np.zeros(length)
    
    erro_df_progressiva_y = np.zeros(length)
    erro_df_regressiva_y = np.zeros(length)
    erro_df_centrada_y = np.zeros(length)
    erro_df_2y = np.zeros(length)
    
    er_df2xy = np.zeros(length)
    
    for i in range(length):
        dfunc_progressiva_x[i] = (func(x0 + deltax[i],y0) - func(x0,y0)) / deltax[i]
        dfunc_regressiva_x[i] = (func(x0,y0) - func(x0-deltax[i],y0)) / deltax[i]
        dfunc_centrada_x[i] = (func(x0 + deltax[i],y0) - func(x0-deltax[i],y0)) / (2*deltax[i])
        dfunc_2cx[i] = (func(x0 + deltax[i],y0) - (2*func(x0,y0)) + func(x0 - deltax[i],y0)) / (deltax[i]**2)
        
        dfunc_progressiva_y[i] = (func(x0, deltay[i]+y0) - func(x0,y0)) / deltay[i]
        dfunc_regressiva_y[i] = (func(x0,y0) - func(x0,y0-deltay[i])) / deltay[i]
        dfunc_centrada_y[i] = (func(x0,  deltay[i]+y0) - func(x0,y0-deltay[i])) / (2*deltay[i])
        dfunc_2centrada_y[i] = (func(x0, deltay[i]+y0) - (2*func(x0,y0)) + func(x0,y0- deltay[i])) / (deltay[i]**2)
        
        dfunc_2centrada_xy[i] = (func(x0+deltax[i], y0+deltay[i]) - func(x0+deltax[i],y0-deltay[i]) - func(x0-deltax[i],y0+deltay[i]) + func(x0-deltax[i],y0-deltay[i])) / (4*deltax[i]*deltay[i])

        
        erro_df_progressivax[i] = abs(dfuncx(x0,y0) - dfunc_progressiva_x[i])/dfuncx(x0,y0)
        erro_df_regressiva_x[i] = abs(dfuncx(x0,y0) - dfunc_regressiva_x[i])/dfuncx(x0,y0)
        erro_df_centrada_x[i] = abs(dfuncx(x0,y0) - dfunc_centrada_x[i])/dfuncx(x0,y0)
        erro_df_2x[i] = abs(Dfuncx(x0,y0) - dfunc_2cx[i])/Dfuncx(x0,y0)
        
        erro_df_progressiva_y[i] = abs(dfuncy(x0,y0) - dfunc_progressiva_y[i])/dfuncy(x0,y0)
        erro_df_regressiva_y[i] = abs(dfuncy(x0,y0) - dfunc_regressiva_y[i])/dfuncy(x0,y0)
        erro_df_centrada_y[i] = abs(dfuncy(x0,y0) - dfunc_centrada_y[i])/dfuncy(x0,y0)
        erro_df_2y[i] = abs(Dfuncy(x0,y0) - dfunc_2centrada_y[i])/Dfuncy(x0,y0)
        
        er_df2xy[i] = abs(Dfuncxy(x0,y0) - dfunc_2centrada_xy[i])/Dfuncxy(x0,y0)
        
    retorno = [
        dfunc_progressiva_x, erro_df_progressivax, dfunc_regressiva_x, erro_df_regressiva_x, dfunc_centrada_x, erro_df_centrada_x, dfunc_2cx, erro_df_2x,
        dfunc_progressiva_y, erro_df_progressiva_y, dfunc_regressiva_y, erro_df_regressiva_y, dfunc_centrada_y, erro_df_centrada_y, dfunc_2centrada_y, erro_df_2y,
        dfunc_2centrada_xy,er_df2xy    
        ]
    return retorno
```

### Exercicio 1 (progressivas, regressivas e centradas)

```python
f = lambda x: 10*np.cos(x)
df = lambda x: - 10* np.sin(x)
Df = lambda x: - 10*np.cos(x)
x0 = np.pi/10
delta = [0.2,0.1, 0.05, 0.025, 0.0125]


retorno = diferenciacao_numerica(x0,f,df,Df,delta)
df_p = retorno[0]
er_p = retorno[1]
df_r = retorno[2]
er_r = retorno[3]
df_c = retorno[4]
er_c = retorno[5]

linhas_tabela = []
for i in range(len(delta)):
    linhas_tabela.append([
        f"{delta[i]:.4f}",
        f"{df_p[i]:.4f}", f"{er_p[i]:.4f}",
        f"{df_r[i]:.4f}", f"{er_r[i]:.4f}",
        f"{df_c[i]:.4f}", f"{er_c[i]:.4f}"
    ])

cabecalhos = [
    "Δx", 
    "Diferença Progressiva", "Erro Relativo", 
    "Diferença Regressiva", "Erro Relativo", 
    "Diferença Centrada", "Erro Relativo"
]

print(tabulate(linhas_tabela, headers=cabecalhos, stralign="center"))
```

### Exercício 3 (progressiva e regressiva)

```python
f = lambda x,y:  np.exp(x * y) + np.sin(x + y)

dfx = lambda x,y: y * np.exp(x * y) + np.cos(x + y)
dfy = lambda x,y: x * np.exp(x * y) + np.cos(x + y)
    
Dfx = lambda x,y: (y**2) * np.exp(x * y) - np.sin(x + y)
Dfy = lambda x,y: (x**2) * np.exp(x * y) - np.sin(x + y)
    
Dfxy = lambda x,y: (1 + x * y) * np.exp(x * y) - np.sin(x + y)
x0 = np.pi/6
y0 = np.pi

delta = [0.2,0.1, 0.05, 0.025] 

retorno = diferenciacao_mista(x0,y0,f,dfx,dfy,Dfx,Dfy,Dfxy,delta,delta)
df_px = retorno[0]
er_px = retorno[1]

df_rx = retorno[2]
er_rx = retorno[3]

df_cx = retorno[4]
er_cx = retorno[5]

df_2x = retorno[6]
er_2x = retorno[7]

df_py = retorno[8]
er_py = retorno[9]

df_ry = retorno[10]
er_ry = retorno[11]

df_cy = retorno[12]
er_cy = retorno[13]

df_2y = retorno[14]
er_2y = retorno[15]

df_2xy = retorno[16]
er_2xy = retorno[17]
```
