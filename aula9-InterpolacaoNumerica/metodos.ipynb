# Interpolação Numérica

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Trapézio (generalizada)

```python
def funcao(x):
    return (np.e**x) * np.cos(x)

# intervalo de discretização
a = 0
b = 1.2
N = 6 # subintervalos
n = N + 1 # número de pontos

h = (b - a) / N

xi = np.linspace(a, b, n)

yi = funcao(xi)

soma = 0
for k in range (0, n):
    if k == 0 or k == n-1:
        soma += yi[k]

    else:
        soma += 2*yi[k]

I_trap = (h/2) * soma

print(f"Valor da integral: {I_trap}")

# para o gráfico
x_fino = np.linspace(a, b, 100)
y_fino = funcao(x_fino)

for k in range (0, N):
    xp = [xi[k], xi[k], xi[k+1], xi[k+1]]
    yp = [0, yi[k], yi[k+1], 0]
    plt.fill(xp, yp, 'b', alpha = 0.3)

for k in range (n):
    plt.plot([xi[k], xi[k]], [0, yi[k]])

plt.plot(xi, yi, 'o')
plt.plot(x_fino, y_fino, 'r-')
plt.show()
```

### $ \frac{1}{3} $ Simpson (generalizado)

```python
def funcao(x):
    return (np.e**x) * np.cos(x)

# intervalo de discretização
a = 0
b = 1.2
N = 3 # subintervalos
n = 2*N + 1 # número de pontos

h = (b - a) / (2 * N)

xi = np.linspace(a, b, n)

yi = funcao(xi)

soma = 0
for k in range (0, n):
    if k == 0 or k == 2*N:
        soma += yi[k]

    elif k % 2 == 1:
        soma += 4 * yi[k]
    else:
        soma += 2 * yi[k]

I_simp_1_3 = (h/3) * soma

print(f"Valor da integral: {I_simp_1_3}")


# para o gráfico
x_fino = np.linspace(a, b, 500)
y_fino = funcao(x_fino)

for k in range (n-1):
    xp = [xi[k], xi[k], xi[k+1], xi[k+1]]
    yp = [0, yi[k], yi[k+1], 0]
    plt.fill(xp, yp, 'b', alpha = 0.3)

for k in range (0, 2 * N, 2):
    plt.plot([xi[k], xi[k]], [0, yi[k]])

plt.plot(xi, yi, 'o')
plt.plot(x_fino, y_fino, 'r-')
plt.show()
```

### $ \frac{3}{8} $ Simpson (generalizada) 

```python
def funcao(x):
    return (np.e**x) * np.cos(x)

# intervalo de discretização
a = 0
b = 1.2
N = 3 # subintervalos
n = 3*N + 1 # número de pontos

h = (b - a) / (3 * N)

xi = np.linspace(a, b, n)

yi = funcao(xi)

soma = 0
for k in range (0, n):
    if k == 0 or k == 3*N:
        soma += yi[k]

    elif k % 3 == 0:
        soma += 2 * yi[k]
    else:
        soma += 3 * yi[k]

I_simp_3_8 = (3 * h / 8) * soma

print(f"Valor da integral: {I_simp_3_8}")


# para o gráfico
x_fino = np.linspace(a, b, 500)
y_fino = funcao(x_fino)

for k in range (n-1):
    xp = [xi[k], xi[k], xi[k+1], xi[k+1]]
    yp = [0, yi[k], yi[k+1], 0]
    plt.fill(xp, yp, 'b', alpha = 0.3)

for k in range (0, 3 * N, 3):
    plt.plot([xi[k], xi[k]], [0, yi[k]])

plt.plot(xi, yi, 'o')
plt.plot(x_fino, y_fino, 'r-')
plt.show()
```
