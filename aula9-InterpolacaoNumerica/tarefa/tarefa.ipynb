## Tarefa

```python
import numpy as np
```

```python
def funcao(x):
    return np.sqrt(1 - x ** 2)
```

```python
area_exata = np.pi / 2.
```

```python
def Trapezio (N):
    a = -1
    b = 1

    n = N + 1 

    h = (b - a) / N

    xi = np.linspace(a, b, n)

    yi = funcao(xi)

    soma = 0
    for k in range (0, n):
        if k == 0 or k == n-1:
            soma += yi[k]

        else:
            soma += 2*yi[k]

    I_trap = (h/2) * soma

    return I_trap
```

```python
def simp_1_3(N):
    a = -1
    b = 1
    
    n = 2*N + 1 

    h = (b - a) / (2 * N)

    xi = np.linspace(a, b, n)

    yi = funcao(xi)

    soma = 0
    for k in range (0, n):
        if k == 0 or k == 2*N:
            soma += yi[k]

        elif k % 2 == 1:
            soma += 4 * yi[k]
        else:
            soma += 2 * yi[k]

    I_simp_1_3 = (h/3) * soma

    return I_simp_1_3
```

```python
def simpson_3_8 (N):
    a = -1
    b = 1

    n = 3*N + 1

    h = (b - a) / (3 * N)

    xi = np.linspace(a, b, n)

    yi = funcao(xi)

    soma = 0
    for k in range (0, n):
        if k == 0 or k == 3*N:
            soma += yi[k]

        elif k % 3 == 0:
            soma += 2 * yi[k]
        else:
            soma += 3 * yi[k]

    I_simp_3_8 = (3 * h / 8) * soma

    return I_simp_3_8
```

```python
def erro_abs(valor, resultado_real):
    return np.abs(resultado_real - valor)
```

### Calculando resultados

```python
i_trapezio6 = Trapezio(6)
erro_trapezio6 = erro_abs(valor = i_trapezio6, resultado_real = area_exata)
i_trapezio12= Trapezio(12)
erro_trapezio12 = erro_abs(valor = i_trapezio12, resultado_real = area_exata)

i_simp_1_3__6 = simp_1_3(6)
erro_simp_1_3__6 = erro_abs(valor = i_simp_1_3__6, resultado_real = area_exata)
i_simp_1_3__12 = simp_1_3(12)
erro_simp_1_3__12 = erro_abs(valor = i_simp_1_3__12, resultado_real = area_exata)

i_simp_3_8__6 = simpson_3_8(6)
erro_simp_3_8__6 = erro_abs(valor = i_simp_3_8__6, resultado_real = area_exata)
i_simp_3_8__12 = simpson_3_8(12)
erro_simp_3_8__12 = erro_abs(valor = i_simp_3_8__12, resultado_real = area_exata)
```
